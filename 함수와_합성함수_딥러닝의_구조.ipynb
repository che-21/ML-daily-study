```python
import numpy as np

# 신경망 한 층 = 행렬곱 + 활성화 함수 (하나의 함수)
def layer1(x):
    return np.maximum(0, 2 * x - 1)    # 선형변환 후 ReLU
def layer2(x):
    return 3 * x + 0.5
# 3층짜리 신경망 = layer2(layer1(입력))처럼 함수를 겹겹이
x = 1.0
h1 = layer1(x)            # 1층 통과
y  = layer2(h1)           # 2층 통과
print("입력 x =", x)
print("1층 출력 =", h1)
print("최종 출력 =", y)
print("\n신경망은 결국 함수를 여러 겹 합성한 것: y = layer2(layer1(x))")
```

```python
import numpy as np

# 활성화 함수 없이 선형변환만 쌓으면?
A = np.array([[2, 0], [0, 3]])     # 1층 (선형)
B = np.array([[1, 1], [0, 1]])     # 2층 (선형)

합쳐진_변환 = A @ B                  # 두 층을 합치면...
print("A @ B =\n", 합쳐진_변환)
print("→ 여전히 '행렬 하나' = 선형변환 하나")
print("층을 아무리 쌓아도 직선밖에 못 그린다!")
```

```python
import numpy as np

z = np.array([-2, -0.5, 0, 0.5, 2])   # 입력값들

sigmoid = 1 / (1 + np.exp(-z))         # 0~1로 눌러 담기 (sigmoid)
relu    = np.maximum(0, z)             # 음수는 0, 양수는 그대로 (ReLu)
tanh    = np.tanh(z)                   # -1~1로 눌러 담기 (tanh)
print("입력 z  :", z)
print("sigmoid :", sigmoid.round(3))
print("relu    :", relu)
print("tanh    :", tanh.round(3))
```

```python
# 연쇄법칙:  dy/dx = (dy/du) × (du/dx)
# y = g(f(x)),  f(x)=2x+1,  g(u)=u²
x = 3
u = 2*x + 1 # f(x)

dg_du = 2 * u                  # g'(u) = 2u
df_dx = 2 # f'(x) = 2
dy_dx = dg_du * df_dx          # 연쇄법칙으로 곱하기
print("u = f(x) =", u)
print("dy/du =", dg_du, ", du/dx =", df_dx)
print("dy/dx = dy/du × du/dx =", dy_dx)
```

```python
# 연쇄법칙 결과가 맞는지 '수치미분'으로 검산
def y(x):
    return (2*x + 1) ** 2 # g(f(x))를 통째로

x = 3
h = 1e-6
근사기울기 = (y(x + h) - y(x - h)) / (2 * h)   # 미분의 정의
print("연쇄법칙 결과 :", 28)
print("수치미분 결과 :", round(근사기울기, 4))
print("일치하는가?", abs(근사기울기 - 28) < 0.001)
```

```python
import numpy as np

# sigmoid의 미분은 최대 0.25 — 층마다 이 값이 곱해진다
최대기울기 = 0.25
print("역전파에서 층을 거슬러 갈수록 기울기가 곱해진다:")
for 층수 in [1, 5, 10, 20]:
    남은기울기 = 최대기울기 ** 층수
    print(f"  {층수:2d}층 통과 → 0.25^{층수} = {남은기울기:.2e}")

print("\n20층이면 기울기가 사실상 0 → 앞쪽 층은 학습이 멈춘다")
```

```python
# 기울기 = 얼마나 가파른가 = 세로 변화 ÷ 가로 변화
def 기울기(가로, 세로):
    return 세로 / 가로

print("완만한 계단: 가로 4칸, 세로 2칸 →", 기울기(4, 2))
print("가파른 계단: 가로 2칸, 세로 6칸 →", 기울기(2, 6))
print("\n숫자가 클수록 가파르다!")
```

```python
# f(x) = x²  (예: 시간 x일 때 이동거리)
def f(x):
    return x ** 2

x = 3 # 이 지점에서의 '순간' 기울기를 알고 싶다
# h(들여다보는 구간)를 점점 줄여 본다
for h in [1, 0.1, 0.01, 0.001]:
    평균기울기 = (f(x + h) - f(x)) / h
    print(f"구간 h={h:<6} → 평균 기울기 = {평균기울기:.4f}")

print("\nh가 0에 가까워질수록 → 6에 딱 붙는다")
print("이 6이 바로 x=3에서의 '순간 기울기(미분값)'")
```

```python
# 골짜기 바닥(가장 낮은 곳) 찾기 = 기울기를 보고 아래로 걷기
def f(x):      return x ** 2 # 골짜기 모양 (바닥은 x=0)
def 기울기(x): return 2 * x       # f의 미분

x = 5.0 # 산 중턱에서 출발
보폭 = 0.2 # 한 번에 얼마나 걸을지 (학습률)
for 걸음 in range(6):
    g = 기울기(x)
    print(f"걸음 {걸음}: 현재 x={x:.3f}, 기울기={g:.3f} → 기울기 반대로 이동")
    x = x - 보폭 * g              # 기울기 '반대' 방향으로 한 발
print(f"\n도착: x={x:.3f} (바닥 x=0에 가까워짐)")
```
